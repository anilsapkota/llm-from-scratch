```python
# we use vector instead of list because vectors are optimized for mathematical operations
# and list for storing 

vec = [1,2,3]
print(vec * 2)

vec_np = np.array(vec)
print(vec_np * 2)
```

```python
#Vectorized operations

vec = np.array([1,2,3])
other = np.array([4,5,6])

#Element-wise operations
print(vec + other)
print(vec * 2)
print(vec ** 2)
```

```python
#aggregations
print(np.sum(vec))
print(np.mean(vec))
print(np.std(vec))
```

## PyTorch

```python
import torch

vec = torch.tensor([1.0,2.0,3.0])
```

```python
type(vec)
```

```python
#shape and indexing
print(vec.shape)
print(vec[0])
print(vec[-1])
print(vec[1:])
```
