```python
def celcius_to_fahrenheit(celcius):
    fahrenheit = (celcius) * 9/5 + 32
    return fahrenheit 
```

```python
#Lambda function 
square = lambda x: x *x 
```

```python
square(5)
```

```python
#Recursive function 
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

```python
#higher order functions 
def apply_twice(func, value):
    return func(func(value))

def add_five(x):
    return x + 5

def sub_five(x):
    return x -5
```

```python
apply_twice(add_five,10)
```

```python
#Generator Functions 

def fibonacci_generator():
    a, b = 0, 1

    while True:
        yield a
        a, b = b, a+b
```

```python
fib = fibonacci_generator()
for _ in range(10):
    print(next(fib))
```

```python
class Adder:
    def __init__(self):
        self.last_result = None 

    def add(self,a,b):
        result = a + b 
        self.last_result = result
        return result 

    def get_last_result(self):
        return self.last_result 
    
```

```python
my_adder = Adder()
```

```python
my_adder.add(2,6)
```

```python
my_adder.get_last_result()
```

```python
another_adder = Adder()
```

```python
another_adder.add(1,9)
```

```python
another_adder.get_last_result()
```

```python
#Encapsulation
my_adder.get_last_result()
```

```python
#Abstraction
my_adder.add(1,9)
```

```python
#Inheritence
class FancyAdder(Adder):  #Here FancyAdder inherits everything from 
    def multiply(self,a,b):
        return a * b 
    
```

```python
my_adder = FancyAdder()
```

```python
my_adder.multiply(2,5)
```

```python
my_adder.add(2,5)
```

```python
my_adder.get_last_result()
```
